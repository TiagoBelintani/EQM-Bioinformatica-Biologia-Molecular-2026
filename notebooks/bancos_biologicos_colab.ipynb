## 1. Por que bancos biológicos existem?

A produção de dados moleculares cresceu a ponto de ser inviável organizar sequências apenas em artigos, planilhas ou arquivos pessoais.

Bancos biológicos permitem armazenar e disponibilizar, de maneira estruturada:

- sequências de DNA e RNA;
- sequências de proteínas;
- genomas;
- dados brutos de sequenciamento;
- informações taxonômicas;
- estruturas moleculares;
- literatura;
- informações sobre projetos e amostras;
- anotações associadas às sequências.

Um banco de dados biológico não é apenas um grande conjunto de letras `A`, `C`, `G` e `T`.

Um registro útil precisa conectar a sequência a informações sobre sua origem e seu contexto.

Por isso, uma ideia central desta aula é:

> **sequência sem contexto é muito menos informativa do que sequência + metadados.**

## 2. NCBI não é sinônimo de GenBank

Esses nomes aparecem juntos com frequência, mas representam coisas diferentes.

### NCBI

O **National Center for Biotechnology Information (NCBI)** mantém diversos recursos de informação biomédica e molecular e fornece ferramentas de busca e acesso a diferentes bases.

### GenBank

O **GenBank** é um arquivo público de sequências de nucleotídeos e das informações associadas a esses registros.

Portanto:

```text
NCBI
 ├── Nucleotide / nuccore
 ├── GenBank
 ├── SRA
 ├── BioProject
 ├── BioSample
 ├── Protein
 ├── Taxonomy
 ├── PubMed
 └── outros recursos
```

Ao pesquisar uma sequência na interface **Nucleotide**, estamos utilizando o sistema de busca do NCBI para acessar registros de sequências nucleotídicas, incluindo registros do GenBank.

Essa distinção será importante nas próximas aulas, porque mais adiante utilizaremos o **SRA** para obter dados brutos de sequenciamento.

## 4. Accession e version: por que o identificador importa?

Um **accession** é um identificador estável atribuído a um registro.

Exemplo:

```text
MK270576
```

Quando aparece:

```text
MK270576.1
```

temos o formato:

```text
ACCESSION.VERSION
```

A parte antes do ponto identifica o registro. A parte depois do ponto indica a versão daquela sequência.

Se a sequência do registro for modificada, o número de versão pode aumentar:

```text
AB123456.1
        ↓
AB123456.2
```

Isso é importante para reprodutibilidade.

Escrever apenas:

```text
usei uma sequência de Grammostola
```

é pouco informativo.

Registrar:

```text
MK270576.1
```

permite apontar para uma versão específica da sequência que foi utilizada.

Ao longo da disciplina, identificadores como accessions, BioProjects e SRA Runs serão tratados como parte da rastreabilidade da análise.

## 5. Anatomia de um registro GenBank

Ao visualizar um registro no formato GenBank, alguns campos aparecem repetidamente.

Nesta prática vamos reconhecer especialmente:

| Campo | O que representa |
|---|---|
| `LOCUS` | resumo técnico do registro, incluindo tamanho e tipo de molécula |
| `DEFINITION` | descrição resumida da sequência |
| `ACCESSION` | identificador do registro |
| `VERSION` | accession associado a uma versão específica da sequência |
| `SOURCE` | fonte biológica |
| `ORGANISM` | organismo e linhagem taxonômica |
| `REFERENCE` | referências ligadas ao registro |
| `FEATURES` | regiões e anotações associadas à sequência |
| `ORIGIN` | início da sequência nucleotídica no formato GenBank |

### Atenção

O registro e a sequência não são a mesma coisa.

```text
REGISTRO
├── identificadores
├── organismo
├── referências
├── features
├── metadados
└── sequência
```

Quando convertermos o registro para FASTA, grande parte desse contexto deixará de aparecer no arquivo.

## 6. GenBank e FASTA respondem a necessidades diferentes

### Formato GenBank

É rico em informações e anotações.

De forma simplificada:

```text
LOCUS
DEFINITION
ACCESSION
VERSION
SOURCE
ORGANISM
REFERENCE
FEATURES
ORIGIN
...
```

### Formato FASTA

É muito mais simples:

```text
>identificador descrição
ATGCGT...
```

Um FASTA possui essencialmente:

1. um **cabeçalho**, iniciado por `>`;
2. a **sequência**.

Essa simplicidade torna FASTA um formato amplamente aceito por programas de bioinformática.

Portanto, não devemos pensar:

```text
GenBank ou FASTA — qual é melhor?
```

A pergunta adequada é:

```text
qual formato é mais apropriado para a tarefa?
```

Para examinar contexto e anotações, GenBank é muito informativo.

Para fornecer uma sequência a muitos programas de análise, FASTA é conveniente.

## 8. Do navegador para uma análise reprodutível

Baixar um FASTA manualmente pelo navegador funciona, mas em um pipeline queremos também saber reproduzir a recuperação computacionalmente.

A lógica será:

```text
accession.version
       ↓
NCBI E-utilities
       ↓
EFetch
       ↓
registro solicitado
       ↓
arquivo salvo no projeto
```

Nesta aula usaremos **EFetch**, uma das E-utilities do NCBI.

Conceitualmente, informamos:

```text
db       = banco que será consultado
id       = identificador do registro
rettype  = formato desejado
retmode  = modo de retorno
```

Para nosso FASTA:

```text
db=nuccore
id=MK270576.1
rettype=fasta
retmode=text
```

Também recuperaremos o mesmo registro no formato GenBank para comparar os arquivos.
